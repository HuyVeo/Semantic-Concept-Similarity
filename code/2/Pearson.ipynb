```python
data_file = r"D:\Semantic-Concept-Similarity\data\BabelNet\Visim-400.txt"
output_file = r"D:\Semantic-Concept-Similarity\data\Similarity\Sim(0-10)\Sim_Vsim_400_GPT.txt"
out=r"D:\Semantic-Concept-Similarity\data\ BabelNet_combine_WordNet\word_chatgpt.txt"
with open(out, "r", encoding="utf-8") as f:
    valid_words = set(line.strip() for line in f if line.strip())

with open(data_file, "r", encoding="utf-8") as f:
    lines = f.readlines()
header = lines[0]
filtered_lines = [header]

for line in lines[1:]:
    parts = line.strip().split("\t")
    if len(parts) >= 2 and parts[0] in valid_words:
        filtered_lines.append(line)

with open(output_file, "w", encoding="utf-8") as f:
    f.writelines(filtered_lines)
```
